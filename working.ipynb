```python
#embending with BERT
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
#encode in embendding work only with string/ list, so we need to convert the data frame to list
```

```python
texts = (
    book_df["Title"].fillna("") + ". " +
    book_df["Authors"].fillna("") + ". "+
    book_df["Description"].fillna("") + ". " +
    book_df["Category"].fillna("") + ". " +
    book_df["Publisher"].fillna("")
)

book_embeddings = model.encode(
    texts.tolist(),
    normalize_embeddings=True
)
```

```python
#embendding to the client request
```

```python
def get_user_embedding(user_query, model):
    return model.encode(user_query, normalize_embeddings=True)
```

```python
#check similarity
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def get_similarity_scores(user_embedding, book_embeddings):
    return cosine_similarity([user_embedding], book_embeddings)[0]
```

```python
#bring top 5 recommends
```

```python
import numpy as np

def get_top_books(book_df, semantic_scores, top_n=5):
    top_idx = np.argsort(semantic_scores)[::-1][:top_n]
    return book_df.iloc[top_idx]
```

```python
#final: call to all the function
```

```python
user_query = input("what do you want?")
user_embedding = get_user_embedding(user_query, model)
semantic_scores = get_similarity_scores(user_embedding, book_embeddings)
top_books = get_top_books(book_df, semantic_scores)

#print(top_books[["Title", "Authors", "Description"]])
top_books
```
