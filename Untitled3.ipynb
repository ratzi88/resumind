```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
#import dataset
job_df = pd.read_csv('fake_job_postings.csv')
#delete useless columns
job_df = job_df.drop(columns=['telecommuting', 'has_company_logo', 'has_questions', 'fraudulent'])
#Reduce the database size
job_df = job_df.head(500)
```

```python
#embendding with BERT
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
#embendding the dataset
```

```python
texts = (
    "job title: " + job_df["title"].fillna("") +
    " job description: " + job_df["description"].fillna("") +
    " job requirements: " + job_df["requirements"].fillna("")
)

job_embeddings = model.encode(
    texts.tolist(),
    normalize_embeddings=True
)
```

```python
def recommend_jobs(user_query, top_n=5):
    #embendding to the user reqest
    user_embedding = model.encode(
        user_query,
        normalize_embeddings=True
    )

    #similarity calculation
    scores = cosine_similarity(
        [user_embedding],
        job_embeddings
    )[0]

    #top_n option
    top_idx = np.argsort(scores)[::-1][:top_n]

    results = job_df.iloc[top_idx].copy()
    results["semantic_score"] = scores[top_idx]

    return results
```

```python
while True:
    query = input("\nWhat job are you looking for? (or 'exit') ")
    if query.lower() == "exit":
        break

    recommendations = recommend_jobs(query)
    print(recommendations[["title", "semantic_score"]])
```
