```python
import pandas as pd

job_df=pd.read_csv('fake_job_postings.csv')
job_df.head(10)
```

```python
job_df=job_df.drop(columns=['telecommuting', 'has_company_logo', 'has_questions', 'fraudulent'])
job_df.head(10)
```

```python
#embendding with BERT
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
job_df=job_df.head(200)
```

```python
#conversion of the necessary information
texts = (
    "job title: "+ job_df["title"] +
    "job description: "+ job_df["description"].fillna("")+
    "job requirements: "+ job_df["requirements"].fillna("")
)
job_embeddings = model.encode(
    texts.tolist(),
    normalize_embeddings=True
)
```

```python
#embendding to the client request
```

```python
def get_user_embedding(user_query, model):
    return model.encode(user_query, normalize_embeddings=True)
```

```python
#check similarity
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def get_similarity_scores(user_embedding, job_embeddings):
    return cosine_similarity([user_embedding], job_embeddings)[0]
```

```python
#chatGPT
```

```python
# -------------------------
# 4️⃣ פונקציית ניקוד מטא־דאטה
# -------------------------
def metadata_score(job_row, user_query):
    score = 0
    user_query = user_query.lower()

    # בדיקה מול employment_type
    if pd.notna(job_row.get("employment_type")):
        if job_row["employment_type"].lower() in user_query:
            score += 1

    # בדיקה מול required_experience
    if pd.notna(job_row.get("required_experience")):
        if job_row["required_experience"].lower() in user_query:
            score += 1

    # בדיקה מול industry
    if pd.notna(job_row.get("industry")):
        if job_row["industry"].lower() in user_query:
            score += 1

    return score
```

```python
# -------------------------
# 5️⃣ קבלת שאילתת המשתמש
# -------------------------
user_query = input("What kind of job are you looking for? ")

user_embedding = get_user_embedding(user_query, model)
semantic_scores = get_similarity_scores(user_embedding, job_embeddings)

# -------------------------
# 6️⃣ הוספת עמודות ניקוד למשרות
# -------------------------
job_df["semantic_score"] = semantic_scores

job_df["metadata_score"] = job_df.apply(
    lambda row: metadata_score(row, user_query),
    axis=1
)

# נרמול ניקוד מטא־דאטה כדי שיהיה באותה סקאלה של הסמנטי
if job_df["metadata_score"].max() > 0:
    job_df["metadata_score"] = job_df["metadata_score"] / job_df["metadata_score"].max()
else:
    job_df["metadata_score"] = 0

# -------------------------
# 7️⃣ שילוב ניקוד סמנטי + מטא־דאטה
# -------------------------
alpha = 0.7  # משקל סמנטי
beta = 0.3   # משקל מטא־דאטה

job_df["final_score"] = alpha * job_df["semantic_score"] + beta * job_df["metadata_score"]

# -------------------------
# 8️⃣ בחירת Top Jobs לפי הציון הסופי
# -------------------------
top_n = 5
top_jobs = job_df.sort_values("final_score", ascending=False).head(top_n)

# הצגת התוצאות
top_jobs[["title", "semantic_score", "metadata_score", "final_score"]]

```
